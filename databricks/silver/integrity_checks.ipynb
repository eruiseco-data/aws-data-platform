```python
%sql
SELECT 'order_items -> orders' AS regla, COUNT(*) AS huerfanos
FROM workspace.silver.order_items oi
LEFT ANTI JOIN workspace.silver.orders o
  ON oi.order_id = o.order_id
UNION ALL
SELECT 'order_items -> products', COUNT(*)
FROM workspace.silver.order_items oi
LEFT ANTI JOIN workspace.silver.products p
  ON oi.product_id = p.product_id
UNION ALL
SELECT 'order_items -> sellers', COUNT(*)
FROM workspace.silver.order_items oi
LEFT ANTI JOIN workspace.silver.sellers p
  ON oi.seller_id = p.seller_id
UNION ALL
SELECT 'order_payments → orders', COUNT(*)
FROM workspace.silver.order_payments op
LEFT ANTI JOIN workspace.silver.orders p
  ON op.order_id = p.order_id
UNION ALL
SELECT 'order_reviews → orders', COUNT(*)
FROM workspace.silver.order_reviews orv
LEFT ANTI JOIN workspace.silver.orders p
  ON orv.order_id = p.order_id
UNION ALL
SELECT 'orders → customers', COUNT(*)
FROM workspace.silver.orders o
LEFT ANTI JOIN workspace.silver.customers cust
  ON o.customer_id = cust.customer_id
UNION ALL
SELECT 'products → product_category_name_translation', COUNT(*)
FROM workspace.silver.products p
LEFT ANTI JOIN workspace.silver.product_category_name_translation pcnt
  ON p.product_category_name = pcnt.product_category_name
WHERE p.product_category_name IS NOT NULL
```
